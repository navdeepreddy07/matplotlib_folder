```python
import pandas as pd
from matplotlib import pyplot as plt

data=pd.DataFrame({"Date":["mon","tue","wed","thu","fri","sat","sun"],
                   "product":["laptop","phone","laptop","tablet","phone","laptop","tablet"],
                   "sales":[60000,50000,75000,25000,30000,90000,45000],
                   "quantity":[2,3,2,4,3,3,5]
                   },index=[1,2,3,4,5,6,7])
```

```python
print(data)
```

```python
# Which day had the highest sales?
print(data.sort_values(by="sales",ascending=False).head(1))
```

```python
# Which product generated the highest total sales?

highest=data.groupby("product")["sales"].sum().sort_values(ascending=False)
print(highest)
```

```python
# Which product had the highest total quantity sold?
quantity=data.groupby("product")[["quantity","sales"]].sum()
print(quantity)
```

```python
# Which day had the highest quantity of products sold?
highest_quantity = data.sort_values(by="quantity",ascending=False)

print(highest_quantity.head(1))
```

```python
# now we are connecting pandas DataFrame to matplotlib
```

```python
dates=data["Date"]
sales=data["sales"]
plt.title("Weekly Sales Analysis")
plt.xlabel("Days")
plt.ylabel("sales")
plt.tick_params(axis="x",color="red")
plt.tick_params(axis="y",color="blue")
plt.plot(dates,sales)
plt.show()
```

```python
total_sales=data.groupby("product")["sales"].sum()
print(total_sales)
plt.plot(total_sales)
```

```python
plt.bar(total_sales.index,total_sales.values)
plt.title("Total Sales by Product")
plt.xlabel("Product Names")
plt.ylabel("Sales amount")
plt.show()
```

```python
quantity=data.groupby("product")["quantity"].sum()
print(quantity)
```

```python
plt.bar(quantity.index,quantity.values)
plt.title("Total Quantity Sold by Product")
plt.xlabel("Product Name ")
plt.ylabel("Quantity sold")
plt.show()
```

```python
plt.plot(data["Date"], data["sales"], label="Sales")
plt.plot(data["Date"], data["quantity"], label="Quantity")

plt.legend()
plt.show()
```
