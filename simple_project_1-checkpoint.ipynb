```python
import pandas as pd
from matplotlib import pyplot as plt

data=pd.DataFrame({"Date":["mon","tue","wed","thu","fri","sat","sun"],
                   "product":["laptop","phone","laptop","tablet","phone","laptop","tablet"],
                   "sales":[60000,50000,75000,25000,30000,90000,45000],
                   "quantity":[2,3,2,4,3,3,5]
                   })
```

```python
print("Jupyter is working")

plt.plot(data["product"],linestyle="--",marker="o",color="black",label="product")
plt.grid()
plt.xlabel("Dates")
plt.ylabel("product")
plt.legend()
plt.bar(data["product"],height=5)
```
