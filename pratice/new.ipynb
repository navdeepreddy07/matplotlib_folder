```python
from matplotlib import pyplot as plt
```

```python
plt.tick_params(axis="x", colors="red")
plt.tick_params(axis="y", colors="blue")
```

```python
dates=[1,2,3,4,5]
sales=[50,60,70,80,90]

```

```python
plt.plot(dates,sales)
plt.tick_params(axis="x", colors="red")
plt.tick_params(axis="y", colors="blue")
plt.title("Sales Analying")
plt.xlabel("dates")
plt.ylabel("sales")
plt.xticks([1,3,5])
plt.yticks([50,70,90])
```

```python
plt.tick_params(axis="x",labelsize=15)
plt.tick_params(axis="x", length=-1)
plt.tick_params(axis="y", width=1)
```

```python
plt.plot(dates,sales)
plt.tick_params(axis="x", colors="red")
plt.tick_params(axis="y", colors="blue")
plt.title("Sales Analying")
plt.xlabel("dates")
plt.ylabel("sales")
plt.xticks([1,3,5])
plt.yticks([50,70,90])
plt.tick_params(axis="x", direction="inout")
```
